```python
def create_decline_features(data):
    data = data.copy()

    metrics = {
        'data_usage': 'data_usage_mb',
        'call_minutes': 'avg_call_minutes',
        'recharge_amount': 'recharge_amount'
    }

    for feature, column in metrics.items():
        comparisons = []

        for month in [6, 7, 8]:
            current = data[f'{column}_{month}']
            next_month = data[f'{column}_{month + 1}']

            decline = (next_month < current).where(
                current.notna() & next_month.notna()
            )

            comparisons.append(decline.astype(float))

        data[f'{feature}_decline_months'] = pd.concat(
            comparisons, axis=1
        ).sum(axis=1, min_count=1)

    return data
```

```python
import optuna
import numpy as np

from xgboost import XGBClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score


cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)
```

```python
def objective(trial):

    params = {
        "n_estimators": trial.suggest_int(
            "n_estimators", 200, 1000, step=100
        ),

        "max_depth": trial.suggest_int(
            "max_depth", 3, 10
        ),

        "learning_rate": trial.suggest_float(
            "learning_rate", 0.01, 0.2, log=True
        ),

        "min_child_weight": trial.suggest_int(
            "min_child_weight", 1, 10
        ),

        "subsample": trial.suggest_float(
            "subsample", 0.6, 1.0
        ),

        "colsample_bytree": trial.suggest_float(
            "colsample_bytree", 0.6, 1.0
        ),

        "gamma": trial.suggest_float(
            "gamma", 0, 5
        ),

        "reg_alpha": trial.suggest_float(
            "reg_alpha", 1e-4, 10, log=True
        ),

        "reg_lambda": trial.suggest_float(
            "reg_lambda", 1e-3, 20, log=True
        ),

        "scale_pos_weight": trial.suggest_float(
            "scale_pos_weight", 1.0, 4.0
        ),

        "random_state": 42,
        "eval_metric": "logloss",
        "n_jobs": 1
    }

    model = XGBClassifier(**params)

    scores = cross_val_score(
        model,
        X_train_sub,
        y_train_sub,
        cv=cv,
        scoring="average_precision",
        n_jobs=1
    )

    return np.mean(scores)
```

```python
study = optuna.create_study(
    direction="maximize",
    study_name="Telecom_XGBoost_Optuna",
    sampler=optuna.samplers.TPESampler(seed=42)
)

study.optimize(
    objective,
    n_trials=50,
    show_progress_bar=True
)
```

```python
print("Best AP Score:", study.best_value)
print("Best Parameters:", study.best_params)


optuna_results = study.trials_dataframe()
optuna_results.sort_values(
    by="value",
    ascending=False
).head(10)
```

```python
best_optuna_model = XGBClassifier(
    **study.best_params,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
)

best_optuna_model.fit(
    X_train_sub,
    y_train_sub
)
```
