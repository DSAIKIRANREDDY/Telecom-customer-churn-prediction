```python

import pandas as pd

df = pd.read_csv("data/interim/telecom_churn.csv")

print("Dataset shape:", df.shape)

print("\nColumn names and data types:")
print(df.dtypes.to_string())

print("\nMissing values:")
print(df.isnull().sum().to_string())

print("\nUnique values in target:")
print(df["Churn"].value_counts(dropna=False))
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nBlank TotalCharges:")
print((df["TotalCharges"].str.strip() == "").sum())

print("\nNon-numeric TotalCharges:")
numeric_total = pd.to_numeric(df["TotalCharges"], errors="coerce")
print(numeric_total.isna().sum())

print("\nDuplicate customer IDs:")
print(df["customerID"].duplicated().sum())
```
